# How evenly do digits appear in π?

### A frequency analysis of 1,000,001 digits

STAT 159 · Lab 11

**Question:** Do digits 0–9 occur at similar rates?

We counted each digit in the supplied notebook. This presentation describes the finite sample, including the initial 3.

## Data and method

- **1,000,001 observations:** the leading 3 plus one million decimal digits.
- Two columns: `digit_position` and `digit`.
- The notebook loads TidyTuesday data, then rebuilds the table from the raw π text file.
- Count each digit and divide by the total. Compare each percentage with **10%**.

Sources: [TidyTuesday, March 24, 2026](https://github.com/rfordatascience/tidytuesday/tree/main/data/2026/2026-03-24) · [Raw π digits](https://github.com/eneko/Pi/blob/master/one-million.txt)

The final frequency table comes from the raw-text cleaning step, which overwrites the initially loaded table. Whitespace is removed and the leading 3 is added back.

## Conclusion and next question

**Digits 0–9 occur at very similar rates in this sample.**

- Similar overall frequencies do not establish independence or randomness.
- A finite sample cannot prove a claim about all digits of π.
